```python
import xarray as xr
```

```python
grid_path = "/cluster/work/math/climate-downscaling/cordex-data/cordex-ALPS-allyear/test/extrapolation/rsds_day_EUR-11_ALADIN63_CNRM-CM5_r1i1p1_rcp85_ALPS_cordexgrid_2090-2099.nc"
orig_data = xr.open_dataset(grid_path)
lons = orig_data.lon.values
lats = orig_data.lat.values

# temporal mean
rsds = orig_data["rsds"]
rsds_mean = rsds.isel(time=0)
# Define your output path
output_file = "../plotting_data/15_rsds.nc"

# Save the time slice
rsds_mean.to_netcdf(output_file)
print(f"File saved successfully to {output_file}")
```
